## Data Visualizations

Now let us explore the dataset visually to understand distributions, missing values, and key patterns.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import ast

# Set style for all plots
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (14, 6)
plt.rcParams["font.size"] = 12

print("✅ Visualization libraries loaded")

# --- 1. Job Position Distribution ---

# Parse positions column
df_pos = df.copy()
df_pos["positions"] = df_pos["positions"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)
df_pos = df_pos.explode("positions").dropna(subset=["positions"])

# Get top 20 positions
top_20 = df_pos["positions"].value_counts().head(20)

plt.figure(figsize=(14, 8))
colors_bar = sns.color_palette("viridis", len(top_20))
bars = plt.barh(range(len(top_20)), top_20.values, color=colors_bar)
plt.yticks(range(len(top_20)), top_20.index)
plt.xlabel("Number of Resumes", fontsize=13)
plt.ylabel("Job Position", fontsize=13)
plt.title("Top 20 Job Positions in Dataset", fontsize=16, fontweight="bold")

for i, (bar, val) in enumerate(zip(bars, top_20.values)):
    plt.text(val + 10, bar.get_y() + bar.get_height() / 2,
             str(val), va="center", fontsize=11)
plt.tight_layout()
plt.show()
print(f"📊 Total unique positions: {df_pos['positions'].nunique()}")
print(f"📊 Total exploded rows: {len(df_pos)}")

# --- 2. Missing Value Analysis ---

missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({"Missing Count": missing, "Percentage": missing_pct})
missing_df = missing_df[missing_df["Missing Count"] > 0].sort_values("Missing Count", ascending=False)

plt.figure(figsize=(14, 8))
colors_miss = sns.color_palette("Reds_r", len(missing_df))
bars2 = plt.barh(range(len(missing_df)), missing_df["Percentage"].values, color=colors_miss)
plt.yticks(range(len(missing_df)), missing_df.index)
plt.xlabel("Missing (%)", fontsize=13)
plt.title("Missing Values by Column", fontsize=16, fontweight="bold")

for i, (bar, val, cnt) in enumerate(zip(bars2, missing_df["Percentage"].values, missing_df["Missing Count"].values)):
    plt.text(val + 0.5, bar.get_y() + bar.get_height() / 2,
             f"{val:.1f}% ({cnt})", va="center", fontsize=10)
plt.tight_layout()
plt.show()
print(f"\n📊 Columns with 0 missing: {sum(missing == 0)}")
print(f"📊 Columns with >50% missing: {sum(missing_pct > 50)}")
print(f"📊 Columns with >90% missing: {sum(missing_pct > 90)}")
```

```python
# --- 3. Top Skills Distribution ---

# Parse skills column (it's stored as string lists)
df_skills = df.copy()
df_skills["skills"] = df_skills["skills"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[") else []
)
df_skills = df_skills.explode("skills").dropna(subset=["skills"])
df_skills["skills"] = df_skills["skills"].str.strip().str.lower()
df_skills = df_skills[df_skills["skills"] != ""]

top_skills = df_skills["skills"].value_counts().head(20)

plt.figure(figsize=(14, 8))
colors_skill = sns.color_palette("Blues_r", len(top_skills))
bars3 = plt.barh(range(len(top_skills)), top_skills.values, color=colors_skill)
plt.yticks(range(len(top_skills)), top_skills.index)
plt.xlabel("Number of Resumes", fontsize=13)
plt.ylabel("Skill", fontsize=13)
plt.title("Top 20 Skills Mentioned in Resumes", fontsize=16, fontweight="bold")

for i, (bar, val) in enumerate(zip(bars3, top_skills.values)):
    plt.text(val + 5, bar.get_y() + bar.get_height() / 2,
             str(val), va="center", fontsize=11)
plt.tight_layout()
plt.show()
print(f"📊 Total unique skills: {df_skills['skills'].nunique()}")

# --- 4. Matched Score Distribution ---

plt.figure(figsize=(12, 5))
sns.histplot(df["matched_score"].dropna(), bins=30, kde=True, color="steelblue")
plt.xlabel("Matched Score", fontsize=13)
plt.ylabel("Frequency", fontsize=13)
plt.title("Distribution of Matched Scores Across Job Postings", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
print(f"📊 Matched Score - Mean: {df['matched_score'].mean():.4f}, Median: {df['matched_score'].median():.4f}")
print(f"📊 Matched Score - Min: {df['matched_score'].min():.4f}, Max: {df['matched_score'].max():.4f}")

# --- 5. Top Job Position Names (from job_position_name column) ---

top_job_names = df["﻿job_position_name"].value_counts().head(15)

plt.figure(figsize=(14, 6))
top_job_names.plot(kind="bar", color=sns.color_palette("mako", len(top_job_names)))
plt.xlabel("Job Position Name", fontsize=13)
plt.ylabel("Count", fontsize=13)
plt.title("Top 15 Job Position Names in Dataset", fontsize=16, fontweight="bold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
print(f"\n📊 All done! EDA complete with visualizations.")
```
